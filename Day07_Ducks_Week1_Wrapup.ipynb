```python
# Game 1 vs LAK
game_1_events = [
    {"event": "shot", "team": "ANA", "player": "Leo Carlsson", "xg": 0.32, "goal": True},
    {"event": "shot", "team": "LAK", "player": "Adrian Kempe", "xg": 0.14, "goal": False},
    {"event": "shot", "team": "ANA", "player": "Cutter Gauthier", "xg": 0.08, "goal": False},
    {"event": "shot", "team": "ANA", "player": "Beckett Sennecke", "xg": 0.41, "goal": True}
]

# Game 2 vs SJS
game_2_events = [
    {"event": "shot", "team": "ANA", "player": "Troy Terry", "xg": 0.19, "goal": False},
    {"event": "shot", "team": "ANA", "player": "Mason McTavish", "xg": 0.27, "goal": True},
    {"event": "hit", "team": "ANA", "player": "Radko Gudas", "xg": 0.0}
]



def summarize_ducks_game(event_list):
  ducks_shots = 0
  ducks_xg = 0
  ducks_goals = 0
  ducks_hits = 0

  for play in event_list:
    if play["event"] == "shot" and play["team"] == "ANA":
      ducks_shots = ducks_shots + 1
      ducks_xg = ducks_xg + play["xg"]
      if play["goal"] == True:
        ducks_goals = ducks_goals + 1
    if play["event"] == "hit":
      ducks_hits = ducks_hits + 1

  return(f"Anaheim Ducks Game Summary: {ducks_shots} Shots | {ducks_goals} Goals | {round(ducks_xg, 2)} Expected goals (xG) | {ducks_hits} Hits")

print("Game 1 ->", summarize_ducks_game(game_1_events))
print("Game 2 ->", summarize_ducks_game(game_2_events))
```
